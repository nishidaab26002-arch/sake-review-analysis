# 01. 口コミのスクレイピング

日本酒口コミサイト「Saketime」の都道府県別ランキングから上位銘柄を取り出し、各銘柄の口コミを収集してCSVに保存します。

| 項目 | 内容 |
|---|---|
| 対象 | 44府県 × 上位10銘柄 × 最大3ページ（1銘柄あたり約30件） |
| 使用ライブラリ | Playwright（ヘッドレスChromium）, BeautifulSoup, pandas |
| 出力 | `saketime_robust_result.csv`（prefecture, brand, maker, review_num, pricerange, review） |

**工夫した点**
- 都道府県ごとに中間CSVを保存し、途中で止まっても続きから再開できる
- 同一銘柄内の口コミ重複を除去
- ページ間・銘柄間に待機時間を入れ、サーバーへの負荷に配慮

次のステップ: `03_parse_review_text.ipynb`（口コミ文字列をユーザー名・評価・スペック等に分解）

## クローラー本体と実行

```python
import asyncio
import re
import random
import os
from playwright.async_api import async_playwright
from bs4 import BeautifulSoup
import pandas as pd

# 最終的な保存先CSVファイルのパス
OUTPUT_PATH = f'{CSV_DIR}/saketime_robust_result.csv'

async def fetch_brand_urls_by_prefecture(page, pref_code, limit=10):
    """
    【処理1】指定された都道府県のランキングページから、上位の銘柄URLを自動収集する関数
    """
    url = f"https://www.saketime.jp/ranking/{pref_code}/"
    print(f"🌐 巡回ルート作成中... [{pref_code.upper()}] ランキングから上位 {limit} 件の銘柄を抽出します。")

    try:
        await page.goto(url, timeout=60000, wait_until='domcontentloaded')
        await asyncio.sleep(3)

        html = await page.content()
        soup = BeautifulSoup(html, 'html.parser')

        brand_urls = []
        for a_tag in soup.find_all('a', href=re.compile(r'^/brands/\d+/$')):
            full_url = f"https://www.saketime.jp{a_tag['href']}"
            if full_url not in brand_urls:
                brand_urls.append(full_url)
            if len(brand_urls) >= limit:
                break

        print(f"  ➔ [{pref_code.upper()}] の巡回対象: {len(brand_urls)} 銘柄が確定しました。")
        return brand_urls

    except Exception as e:
        print(f"  ⚠️ [{pref_code.upper()}] ランキングの自動取得に失敗しました: {e}")
        return []


async def scrape_brand_reviews(page, brand_url, max_pages=3):
    """
    【処理3〜5】1つの銘柄から、指定ページ数分のプレーン口コミを「重複なく」抽出する関数
    """
    brand_data_rows = []
    base_url = brand_url

    for page_idx in range(1, max_pages + 1):
        current_url = base_url if page_idx == 1 else f"{base_url}page:{page_idx}/"
        print(f"   ↳ [{page_idx} / {max_pages} ページ目] アクセス中... ({current_url})")

        try:
            await page.goto(current_url, timeout=60000, wait_until='domcontentloaded')

            try:
                await page.wait_for_selector('#review', timeout=10000)
            except Exception:
                print(f"     ⚠️ ページ {page_idx} で口コミエリアの読み込みに失敗しました。スキップします。")
                continue

            await asyncio.sleep(3) # レンダリング待機
            await page.evaluate("window.scrollTo(0, document.body.scrollHeight / 3);")
            await asyncio.sleep(2)

            html_content = await page.content()
            soup = BeautifulSoup(html_content, 'html.parser')
            page_text = soup.get_text()

            # 基本情報の抽出
            brand_element = soup.find('h1')
            brand_name = brand_element.get_text(strip=True).split('|')[0].strip() if brand_element else "不明"

            pref_element = soup.find('a', class_='pref_link')
            prefecture = pref_element.get_text(strip=True) if pref_element else "不明"

            maker_element = soup.find('a', href=re.compile(r'#maker$'))
            if not maker_element:
                brandinfo_p = soup.find('p', class_='brandinfo')
                if brandinfo_p:
                    a_tags = brandinfo_p.find_all('a')
                    if len(a_tags) >= 2:
                        maker_element = a_tags[1]
            maker_info = maker_element.get_text(strip=True) if maker_element else "不明"

            review_num = "不明"
            review_num_element = soup.find('p', class_='review_num')
            if review_num_element:
                review_num_match = re.search(r'(\d+)', review_num_element.get_text(strip=True))
                if review_num_match:
                    review_num = review_num_match.group(1) + "件"

            pricerange = "記載なし"
            price_match = re.search(r'(¥[\d,]+\s*～\s*¥[\d,]+|[\d,]+円\s*～\s*[\d,]+円)', page_text)
            if price_match:
                pricerange = re.sub(r'\s+', ' ', price_match.group(1)).strip()

            # 口コミの抽出
            review_container = soup.find(id='review')
            page_reviews_count = 0

            if review_container:
                ul_element = review_container.find('ul')
                if ul_element:
                    lis = ul_element.find_all('li', recursive=False)

                    for li in lis:
                        text = li.get_text(strip=True)

                        if 50 < len(text) < 1500:
                            clean_text = re.sub(r'^(recommendピックアップレビュー)', '', text)

                            is_duplicate = False
                            for r in brand_data_rows:
                                existing_review = r['review']
                                if clean_text in existing_review or existing_review in clean_text:
                                    is_duplicate = True
                                    break
                                if clean_text[:30] in existing_review:
                                    is_duplicate = True
                                    break

                            if not is_duplicate:
                                brand_data_rows.append({
                                    "prefecture": prefecture,
                                    "brand": brand_name,
                                    "maker": maker_info,
                                    "review_num": review_num,
                                    "pricerange": pricerange,
                                    "review": clean_text
                                })
                                page_reviews_count += 1

            print(f"     ➔ 重複のない有効な口コミを {page_reviews_count} 件回収しました。")

            if page_idx < max_pages:
                wait_time = random.uniform(10.0, 15.0)
                print(f"     ⏱️ 次のページへ進む前に {wait_time:.2f} 秒待機...")
                await asyncio.sleep(wait_time)

        except Exception as e:
            print(f"      ❌ ページ {page_idx} の取得中にエラーが発生しました: {e}")
            continue

    return brand_data_rows


async def main_prefecture_crawler(pref_list, max_brands_per_pref=10, max_pages_per_brand=3):
    """
    【メイン管制塔】都道府県分割保存・自動再開機能付き
    """
    output_dir = os.path.dirname(OUTPUT_PATH)
    os.makedirs(output_dir, exist_ok=True)

    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        context = await browser.new_context(
            user_agent='Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36',
            viewport={'width': 1280, 'height': 800}
        )
        page = await context.new_page()

        # 都道府県のループ
        for pref in pref_list:
            # 都道府県ごとの個別一時保存ファイルパス
            pref_output_path = os.path.join(output_dir, f"tmp_raw_{pref}.csv")

            print(f"\n==================================================")
            print(f"🗺️ 【地域巡回】都道府県「{pref.upper()}」の調査")
            print(f"==================================================")

            # 🔥【レジューム機能】すでに該当県のファイルが存在すれば丸ごとスキップ
            if os.path.exists(pref_output_path):
                print(f"⏭️ [{pref.upper()}] は既にデータを取得済みのため、スキップして次の県へ進みます。")
                continue

            target_urls = await fetch_brand_urls_by_prefecture(page, pref_code=pref, limit=max_brands_per_pref)
            pref_all_reviews = []

            for index, url in enumerate(target_urls, start=1):
                print(f"\n🚀 [銘柄 {index} / {len(target_urls)}] 巡回開始：{url}")

                brand_reviews = await scrape_brand_reviews(page, url, max_pages=max_pages_per_brand)
                if brand_reviews:
                    pref_all_reviews.extend(brand_reviews)
                    print(f"💾 銘柄『{brand_reviews[0]['brand']}』の口コミ {len(brand_reviews)} 件をメモリに保持しました。")
                else:
                    print(f"⚠️ 有効な口コミが見つかりませんでした。")

                if index < len(target_urls):
                    nest_wait = random.uniform(15.0, 30.0)
                    print(f"🤖【ブロック徹底防御】次の銘柄へ移動する前に {nest_wait:.2f} 秒間休憩します...")
                    await asyncio.sleep(nest_wait)

            # 1つの都道府県が完全に終わったらCSVに書き出し（セーフティガード）
            if pref_all_reviews:
                df_pref = pd.DataFrame(pref_all_reviews)
                df_pref.to_csv(pref_output_path, index=False, encoding='utf-8-sig')
                print(f"📂 [{pref.upper()}] の全 {len(pref_all_reviews)} 件のデータを中間ファイルとして保存しました。")

            print(f"\n☕ [{pref.upper()}] の全処理が完了。次の都道府県に移る前に 20 秒休憩します...")
            await asyncio.sleep(20)

        await browser.close()

    # =========================================================
    # 分割されたCSVを1つに結合
    # =========================================================
    print("\n🔗 ==================================================")
    print("🔗 全都道府県データの結合一括処理を開始します...")
    print("🔗 ==================================================")

    combined_dfs = []
    for pref in pref_list:
        pref_output_path = os.path.join(output_dir, f"tmp_raw_{pref}.csv")
        if os.path.exists(pref_output_path):
            combined_dfs.append(pd.read_csv(pref_output_path))

    if combined_dfs:
        df_final_all = pd.concat(combined_dfs, ignore_index=True)
        df_final_all.to_csv(OUTPUT_PATH, index=False, encoding='utf-8-sig')
        print(f"\n🎉 祝・全巡回＆統合完了！ 最終成果物（プレーンデータ）はこちら：\n➡️ {OUTPUT_PATH}")
        print(f"📊 総取得口コミ数: {len(df_final_all)} 件")

        # 中間ファイルのクリーンアップ（任意。ポートフォリオの形跡として残したければコメントアウトしてください）
        # for pref in pref_list:
        #     pref_output_path = os.path.join(output_dir, f"tmp_raw_{pref}.csv")
        #     if os.path.exists(pref_output_path): os.remove(pref_output_path)
    else:
        print("⚠️ 結合対象のデータが見つかりませんでした。")


# 💡 【本番仕様：沖縄・鹿児島・宮崎を除く44府県の完全網羅リスト】
target_44_prefectures = [
    'hokkaido', 'aomori', 'iwate', 'miyagi', 'akita', 'yamagata', 'fukushima',
    'ibaraki', 'tochigi', 'gunma', 'saitama', 'chiba', 'tokyo', 'kanagawa',
    'niigata', 'toyama', 'ishikawa', 'fukui', 'yamanashi', 'nagano', 'gifu',
    'shizuoka', 'aichi', 'mie', 'shiga', 'kyoto', 'osaka', 'hyogo', 'nara',
    'wakayama', 'tottori', 'shimane', 'okayama', 'hiroshima', 'yamaguchi',
    'tokushima', 'kagawa', 'ehime', 'kochi', 'fukuoka', 'saga', 'nagasaki',
    'kumamoto', 'oita'
]

# ⚙️ プランA条件でクローラー起動！
await main_prefecture_crawler(
    pref_list=target_44_prefectures,
    max_brands_per_pref=10,      # 各県上位10銘柄
    max_pages_per_brand=3        # 1銘柄最大3ページ（約30件）
)
```
